````python
from pathlib import Path
from typing import Any, Dict, Optional, List
import sys
import importlib

def evaluate_policy(report: Dict[str, Any], policy: Any = None) -> Dict[str, Any]:
    """Evaluates an AI-BOM report with flexible schema validation."""
    if not isinstance(report, dict): return {'passed': False, 'violations': ['Invalid report type']}
    required_keys = {'spec_version', 'metadata', 'components'}
    if not required_keys.issubset(set(report.keys())):
        missing = required_keys - set(report.keys())
        return {'passed': False, 'violations': [f'Missing required keys: {missing}']}

    violations = []
    policy_rules = policy.get('rules', []) if policy else []
    components = report.get('components', [])

    for rule in policy_rules:
        if rule.get('type') == 'disallow':
            target = rule.get('target')
            if any(c.get('name') == target for c in components):
                violations.append(f'Disallowed component: {target}')

    return {'passed': len(violations) == 0, 'violations': violations}

# Persist with explicit content to avoid docstring concatenation errors
with open('/content/AI-BOM-Inspector-Enhanced/src/aibom_inspector/policy.py', 'w') as f:
    f.write("""from typing import Any, Dict, Optional, List

def evaluate_policy(report: Dict[str, Any], policy: Any = None) -> Dict[str, Any]:
    \"\"\"Evaluates an AI-BOM report with flexible schema validation.\"\"\"
    if not isinstance(report, dict): return {'passed': False, 'violations': ['Invalid report type']}
    required_keys = {'spec_version', 'metadata', 'components'}
    if not required_keys.issubset(set(report.keys())): return {'passed': False, 'violations': ['Missing required keys']}
    violations = []
    policy_rules = policy.get('rules', []) if policy else []
    for rule in policy_rules:
        if rule.get('type') == 'disallow':
            target = rule.get('target')
            if any(c.get('name') == target for c in report.get('components', [])):
                violations.append(f'Disallowed component: {target}')
    return {'passed': len(violations) == 0, 'violations': violations}
""")
````

````python
import sys
import subprocess
import os
from typing import Any, List, Optional, Set, Dict

def trace_python(target_script: str, args: Optional[List[str]] = None) -> Dict[str, Any]:
    """Traces a Python script execution to discover AI/ML library usage."""
    discovered_imports = set()
    frameworks = ['torch', 'tensorflow', 'keras', 'sklearn', 'pandas', 'numpy', 'transformers', 'diffusers', 'jax']
    target_path_abs = os.path.abspath(target_script)
    script_args = args if args else []

    # Use a robust wrapper that checks both sys.modules and importlib
    wrapper = f"""
import sys
import runpy
import importlib.util
import os

target = {target_path_abs!r}
frameworks = {frameworks!r}

try:
    sys.argv = [target] + {script_args!r}
    runpy.run_path(target, run_name='__main__')
finally:
    for lib in frameworks:
        if lib in sys.modules or importlib.util.find_spec(lib) is not None:
            print(f'DETECTED_IMPORT:{lib}', flush=True)
"""

    try:
        env = os.environ.copy()
        env['PYTHONPATH'] = os.pathsep.join(sys.path)

        process = subprocess.Popen(
            [sys.executable, '-u', '-c', wrapper],
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True,
            env=env
        )
        stdout, stderr = process.communicate(timeout=30)

        for line in stdout.splitlines():
            if line.startswith('DETECTED_IMPORT:'):
                lib = line.split(':', 1)[1].strip()
                discovered_imports.add(lib)

        return {
            'status': 'success' if process.returncode == 0 else 'warning',
            'discovered_imports': sorted(list(discovered_imports)),
            'stdout': stdout,
            'stderr': stderr
        }
    except Exception as e:
        return {'status': 'error', 'message': str(e), 'discovered_imports': []}

# Persist to file system for internal module consistency
with open('/content/AI-BOM-Inspector-Enhanced/src/aibom_inspector/runtime_trace.py', 'w') as f:
    f.write(r"""import sys
import subprocess
import os
from typing import Any, List, Optional, Set, Dict

def trace_python(target_script: str, args: Optional[List[str]] = None) -> Dict[str, Any]:
    discovered_imports = set()
    frameworks = ['torch', 'tensorflow', 'keras', 'sklearn', 'pandas', 'numpy', 'transformers', 'diffusers', 'jax']
    target_path_abs = os.path.abspath(target_script)
    script_args = args if args else []
    wrapper = f"import sys; import runpy; import importlib.util; target={target_path_abs!r}; sys.argv=[target]+{script_args!r}; try: runpy.run_path(target, run_name='__main__'); finally: [print('DETECTED_IMPORT:'+lib, flush=True) for lib in {frameworks!r} if lib in sys.modules or importlib.util.find_spec(lib)]"
    try:
        env = os.environ.copy()
        env['PYTHONPATH'] = os.pathsep.join(sys.path)
        res = subprocess.run([sys.executable, '-u', '-c', wrapper], capture_output=True, text=True, env=env, timeout=30)
        for line in res.stdout.splitlines():
            if line.startswith('DETECTED_IMPORT:'): discovered_imports.add(line.split(':')[1].strip())
        return {'status': 'success', 'discovered_imports': sorted(list(discovered_imports)), 'stdout': res.stdout}
    except Exception as e: return {'status': 'error', 'message': str(e), 'discovered_imports': []}
""")
````

````python
from pathlib import Path
from typing import Any, Dict, Optional, List
import sys

def evaluate_policy(report: Dict[str, Any], policy: Any = None) -> Dict[str, Any]:
    """Evaluates an AI-BOM report with flexible schema validation."""
    if not isinstance(report, dict):
        return {'passed': False, 'violations': [f'Invalid report type: {type(report)}']}

    required_keys = {'spec_version', 'metadata', 'components'}
    found_keys = set(report.keys())

    if not required_keys.issubset(found_keys):
        missing = required_keys - found_keys
        return {
            'passed': False,
            'violations': [f'Invalid report payload schema: Missing {missing}']
        }

    return {'passed': True, 'violations': []}
````
